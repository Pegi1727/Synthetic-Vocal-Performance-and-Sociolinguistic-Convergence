This notebook operationalizes the comparative framework resolving the $N=1$ limitation of the original submission by benchmarking the synthetic performance against human multi-vocal collaboration.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.dpi'] = 300
PALETTE = {'indigo': '#1A237E', 'gold': '#FFB300', 'pink': '#D81B60', 'turquoise': '#00ACC1'}
```

## 1. Load Comparative Master Dataset & Table 2 Reconstructed

```
df_master = pd.read_csv('/mnt/data/comparative_acoustic_master.csv')
df_table2 = pd.read_csv('/mnt/data/table2_comparative_reconstructed.csv')

print("=== Table 2 Reconstructed: Human vs. Synthetic Comparative Acoustic Summary ===")
display(df_table2)
```

## 2. Voice Quality Dynamics (Jitter, Shimmer, HNR)
Analyzing voice quality metrics from the human baseline to contrast with synthetic acoustic stability.

```
human_data = df_master[df_master['Track'] == 'Human'].sort_values('Segment')
synth_data = df_master[df_master['Track'] == 'Synthetic'].sort_values('Segment')

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
metrics = [('jitter_local_mean', 'Local Jitter (%)'), ('shimmer_local_mean', 'Local Shimmer (%)'), ('hnr_mean', 'HNR (dB)')]

for ax, (metric, title) in zip(axes, metrics):
    if metric in human_data.columns:
        ax.bar(human_data['Segment'], human_data[metric], color=PALETTE['indigo'], alpha=0.85, edgecolor='black')
        ax.set_title(title, fontweight='bold', color=PALETTE['indigo'], fontsize=11)
        ax.set_xlabel('Segment')
        ax.grid(axis='y', linestyle=':', alpha=0.7)

plt.tight_layout()
plt.show()
```

## 3. Multi-Track Pitch Dynamics Comparison
Plotting comparative $F_0$ trajectories over temporal segments to verify decoupling and variance dynamics.

```
plt.figure(figsize=(10, 4.5))
plt.plot(human_data['Segment'], human_data['f0_var'], label='Human Track (TaTaTa) F0 Var', color=PALETTE['pink'], marker='s', lw=2.2)
plt.plot(synth_data['Segment'], synth_data['f0_var'], label='Synthetic Track F0 Var', color=PALETTE['turquoise'], marker='o', lw=2.2)
plt.title("Comparative Pitch Variance Across Temporal Segments (S1–S4)", fontsize=12, fontweight='bold', color=PALETTE['indigo'])
plt.xlabel("Segment")
plt.ylabel("F0 Variance (Hz²)")
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(frameon=True)
plt.tight_layout()
plt.show()
```
